```python
import tensorflow as tf

MODEL_PATH = "/content/drive/MyDrive/wafer_xai2_model.keras"

model = tf.keras.models.load_model(MODEL_PATH)
print("Loaded model")
```

```python
import tensorflow as tf

converter = tf.lite.TFLiteConverter.from_keras_model(model)

# ---- KEY SETTINGS (guaranteed to run) ----
converter.optimizations = [tf.lite.Optimize.DEFAULT]
converter.target_spec.supported_types = [tf.float16]   # <-- BIG FIX

tflite_model = converter.convert()

TFLITE_PATH = "/content/drive/MyDrive/wafer_2mb.tflite"
with open(TFLITE_PATH, "wb") as f:
    f.write(tflite_model)

print("Saved:", TFLITE_PATH)


```
